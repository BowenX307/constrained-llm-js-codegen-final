# IterGen v2 and v3 on G4: pass@1 on all 397 tasks, rerunning only the tasks that can change

**Runtime: G4** (required; the notebook stops on any other GPU). Then **Runtime > Run all**. Nothing to edit.
**About 2 hours.** Results go straight to Drive. If the runtime disconnects, run it again: finished tasks are skipped.
If the first cell asks for a restart (transformers version), do Runtime > Restart session, then Run all again.

Code commit 9e192a0:
- **v2** = the three fixes (ace68e3, fa1fa66, 1c9a4c5): modes `itergen_off` and `itergen_on`
- **v3** = v2 + the helper repair (3836cb8, 9e192a0): mode `itergen_on_repair`

**Why only some tasks.** On the same GPU, IterGen writes exactly what unconstrained decoding writes until it
intervenes (a mask, or a backtrack with the checks on). The fixes only change what happens at an intervention.
So a task where the report's run (v1, G4) never intervened, and never hit the time limit, gets the same output
now, and keeps the report's result. This notebook reruns only the others: every task the report's run masked,
backtracked or cut by the time limit, every task the offline replay says the fixed code intervenes on, and
mbpp_563, 725, 771 (the brace fix). That is 66 / 73 tasks with the checks off and
76 / 80 with them on (Qwen2.5 / Qwen3.5). As a check, it also reruns 6 tasks per mode that should not change and
compares their output with the report's, character by character.

Output on Drive: `final/<model>/<mode>_G4_targeted_9e192a0/` and `final/targeted_summary_9e192a0.json`.

```python
def restore_mask_store(model):
    grammar = f'{REPO}/generators/grammars/javascript.lark'
    probe = subprocess.run([sys.executable, '-c',
        'import syncode.common as c; from syncode.parsers.grammars import Grammar\n'
        f'tok = c.load_tokenizer({model!r}); g = Grammar({grammar!r})\n'
        'print("MS_PATH=" + c.SYNCODE_CACHE + "mask_stores/" + type(tok).__name__ + '
        '"/grammar_strict_" + g.hash() + "_" + str(tok.vocab_size) + ".pkl")'],
        env=ENV, cwd=REPO, capture_output=True, text=True)
    lines = [l for l in probe.stdout.splitlines() if l.startswith('MS_PATH=')]
    if probe.returncode != 0 or not lines:
        raise RuntimeError('could not resolve the mask store path:\n' + probe.stderr[-2000:])
    path = lines[-1][len('MS_PATH='):]
    zst = f'{MS_DRIVE}/{os.path.basename(path)}.zst'
    os.makedirs(os.path.dirname(path), exist_ok=True)
    if not (os.path.exists(path) and os.path.getsize(path) > 100 * 2**20):
        if not os.path.exists(zst):
            raise RuntimeError(f'mask store not on Drive: {zst}')
        r = subprocess.run(['zstd', '-q', '-d', '-f', '-T0', zst, '-o', path + '.part'], capture_output=True, text=True)
        if r.returncode != 0:
            raise RuntimeError('the copy on Drive is unreadable: ' + r.stderr[-300:])
        os.replace(path + '.part', path)
    print('mask store ready:', os.path.basename(path))

import random

def status_of(run_dir):
    p = f'{run_dir}/eval.jsonl'
    return {json.loads(l)['name']: json.loads(l)['status'] for l in open(p)} if os.path.exists(p) else {}

quiet = ('Warning', 'warn(', 'Loading weights', 'torch_dtype', 'httpx')
ALL = [json.loads(l)['name'] for l in open(f'{REPO}/datasets/js_prompts_mbpp.jsonl') if l.strip()]
assert len(ALL) == 397, len(ALL)
# mode -> the report run it is compared with and whose task set it reruns
MODES = [('itergen_on_repair', 'itergen_on'), ('itergen_on', 'itergen_on'), ('itergen_off', 'itergen_off')]

def run_mode(model, mode, tasks):
    slug = model.split('/')[-1]
    os.makedirs('/content/work', exist_ok=True)
    tasks_file = f'/content/work/{slug}_{mode}_tasks.txt'
    open(tasks_file, 'w').write('\n'.join(tasks) + '\n')
    out = f'{DRIVE_ROOT}/final/{slug}/{mode}_G4_targeted_{COMMIT[:7]}'
    t0 = time.time()
    r = subprocess.run([sys.executable, 'run.py', '--model', model, '--mode', mode, '--out', out,
                        '--tasks_file', tasks_file], cwd=REPO, env=ENV,
                       text=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT)
    print('\n'.join(l for l in r.stdout.splitlines() if not any(s in l for s in quiet))[-800:])
    if r.returncode != 0:
        raise RuntimeError(f'{slug} {mode}: run.py failed')
    subprocess.run([sys.executable, 'evaluate.py', out], cwd=REPO, env=dict(ENV, TZ='UTC'),
                   capture_output=True, text=True)
    print(f'{slug} {mode}: {len(tasks)} tasks, {(time.time() - t0) / 60:.0f} min')
    return out

summary = []
try:
    for model, sets in SETS.items():
        slug = model.split('/')[-1]
        print(f'\n######## {slug}')
        restore_mask_store(model)
        unc = status_of(f'{DRIVE_ROOT}/final/{slug}/unconstrained_{REPORT_TAG}')
        for mode, base in MODES:
            touched = sets[base]
            controls = random.Random(0).sample(sorted(set(ALL) - set(touched)), CONTROLS)
            out = run_mode(model, mode, touched + controls)
            rep_dir = f'{DRIVE_ROOT}/final/{slug}/{base}_{REPORT_TAG}'
            rep, now = status_of(rep_dir), status_of(out)
            def completion(d, n):
                p = f'{d}/{n}.json'
                return json.load(open(p))['completion'] if os.path.exists(p) else None
            differ = [n for n in controls if completion(out, n) is None or completion(out, n) != completion(rep_dir, n)]
            same = len(controls) - len(differ)
            if differ:
                print('   CONTROL TASKS THAT DIFFER FROM THE REPORT:', differ)
            combined = {n: (now.get(n) if n in touched else rep.get(n)) for n in ALL}
            changed = [(n, rep.get(n), now.get(n)) for n in touched if rep.get(n) != now.get(n)]
            row = dict(model=slug, mode=mode, report_run=base, rerun=len(touched),
                       passed=sum(v == 'OK' for v in combined.values()),
                       report=sum(rep.get(n) == 'OK' for n in ALL), unconstrained=sum(unc.get(n) == 'OK' for n in ALL),
                       controls_identical=f'{same}/{len(controls)}',
                       gained=[n for n, a, b in changed if b == 'OK'], lost=[n for n, a, b in changed if a == 'OK'])
            summary.append(row)
            print(f'== {slug} {mode}: pass@1 {row["passed"]}/397 (report {base}: {row["report"]}, unconstrained: '
                  f'{row["unconstrained"]}); controls identical to the report: {row["controls_identical"]}')
            print(f'   now pass, report failed: {row["gained"]}')
            print(f'   now fail, report passed: {row["lost"]}')
    json.dump(summary, open(f'{DRIVE_ROOT}/final/targeted_summary_{COMMIT[:7]}.json', 'w'), indent=1)
    print('\n==================== pass@1 on all 397 tasks, G4 ====================')
    print(f'{"model":22} {"mode":18} {"unconstr.":>9} {"v1 report":>9} {"now":>5}  controls')
    for r in summary:
        print(f'{r["model"]:22} {r["mode"]:18} {r["unconstrained"]:>9} {r["report"]:>9} {r["passed"]:>5}  {r["controls_identical"]}')
    print('\n(itergen_off and itergen_on are v2, itergen_on_repair is v3; "v1 report" is the report run they are compared with)')
    print('ALL DONE')
finally:
    if AUTO_DISCONNECT:
        from google.colab import drive, runtime
        drive.flush_and_unmount()
        time.sleep(30)
        runtime.unassign()
```
